## 슬라이싱 (Slicing)
- 형식 : [start : end : step]
    - start : 시작 위치
    - end : 종료 위치 >> end는 포함하지 않음 즉, end 위치 전까지 가져옵니다.
    - step : 몇 칸씩 이동할지 지정

```python
b="GoodMorning"
print(b[0:4:1])      
print(b[0:4])        # step 1, 생략 가능
print(b[:4])         # satrt 0, step 1, 생략 가능  
```

```python
print(b[4:11:1])
print(b[4:11])       # step 1, 생략 가능
print(b[4:])         # end 맨끝, step 1, 생략 가능  
```

```python
print(b[::2])
print(b[2::])
print(b[:4:2])
print(b[::])
```

# Quiz02

- BMI(체질량지수) 계산 프로그램
    - 170 키(cm)와 65 몸무게(kg)를 변수에 저장
    - BMI = 몸무게(kg) ÷ (키(m) × 키(m))
    - 추가 조건)
       - BMI < 18.5 : 저체중
       - 18.5 이상 ~ 23 미만 : 정상
       - 23 이상 ~ 25 미만 : 과체중
       - 25 이상 : 비만

```python
height=170      
weight=65  
bmi=weight / ((height / 100) ** 2)
print(bmi)

if bmi < 18.5:
    print("저체중")
elif bmi < 23:
    print("정상")
elif bmi < 25:
    print("과체중")
else:
    print("비만")
```

# 함수
- 입력 값 : 매개변수(parameter) / 인수(argument)

```python
# CASE 1) 함수 정의
def sub():
    print("sub 함수입니다.")
```

```python
# 함수 호출
for i in range(5):
    sub()
```

```python
# CASE 2) 함수 정의
def apple(su):
    for i in range(su):
        print("*", end=' ')
    print()
```

```python
# 함수 호출
apple(5)
apple(10)
```

```python
#  CASE 3) 함수 정의
def hap(a, b):
    return a + b

def mul(a, b):
    return a * b
```

```python
# 함수 호출
a=hap(10, 20)
b=mul(55, 66)
print(a, b)
```
