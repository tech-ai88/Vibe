## 공공 포털 사이트
- https://www.data.go.kr/

- 프롬프트
    - 아래의 내용을 참조하여 기상청 api 날씨 정보 파이썬 코드를 작성해줘
        - url : "http://apis.data.go.kr/1360000/VilageFcstInfoService_2.0/getUltraSrtNcst?serviceKey=인증키&numOfRows=10&pageNo=1&base_date=20210628&base_time=0600&nx=55&ny=127"
        - 인증키는 발급 받았음

```python
import requests
from datetime import datetime
from urllib.parse import unquote


# 기상청 API 인증키
# SERVICE_KEY =unquote("wdkT93ooCcYPPWCPWIluRHhUDdOeVrWNekh%2B%2BlQ5ctI5Z9zLH%2FJjV6cblkDmzlqOicnFBWwV5txBotvhrQyrjg%3D%3D")

# ==========================================
# 1. 인증키 입력 : 인코딩 >> 디코딩 필요 >> 언쿼터 
# ==========================================
SERVICE_KEY = unquote("wdkT93ooCcYPPWCPWIluRHhUDdOeVrWNekh%2B%2BlQ5ctI5Z9zLH%2FJjV6cblkDmzlqOicnFBWwV5txBotvhrQyrjg%3D%3D")

# ==========================================
# 2. API 주소
# ==========================================
url = "http://apis.data.go.kr/1360000/VilageFcstInfoService_2.0/getUltraSrtNcst"

# ==========================================
# 3. 현재 날짜
# ==========================================
now = datetime.now()

base_date = now.strftime("%Y%m%d")

# 초단기실황은 정시 기준으로 조회
base_time = now.strftime("%H00")

# ==========================================
# 4. 서울 지역 예시
#    nx=55, ny=127
# ==========================================
params = {
    "serviceKey": SERVICE_KEY,
    "numOfRows": 10,
    "pageNo": 1,
    "dataType": "JSON",
    "base_date": base_date,
    "base_time": base_time,
    "nx": 55,
    "ny": 127
}

# ==========================================
# 5. API 요청
# ==========================================
response = requests.get(url, params=params)

print("HTTP 상태코드:", response.status_code)

# ==========================================
# 6. 결과 확인
# ==========================================
if response.status_code == 200:

    data = response.json()

    items = data["response"]["body"]["items"]["item"]

    print("\n===== 현재 날씨 =====")

    for item in items:
        category = item["category"]
        value = item["obsrValue"]

        print(category, ":", value)

else:
    print("API 호출 실패")
    print(response.text)
```

# ifream
- 1. Python과 Selenium을 사용하여 아래의 조건에 맞게 코드를 작성해줘
    - 1-1) ifream 요소에 아이디는 "_IframeBannerRight_tgtLREC"이며, 이를 i_fream 변수에 할당해줘
    - 1-2) i_fream변수에 할당된 요소안 새로운 html 문서안 요소중 class명은 "content_info"이며, 이를 content 변수에 할당해줘
    - 1-3) content 변수에 할당된 요소안에는 각기 다른 자식 태그가 3개 있는데, 자식 태그의 내용을 출력해줘

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time

# 브라우저 실행
driver = webdriver.Chrome()

# 페이지 접속
driver.get("https://finance.naver.com/item/main.naver?code=005930")

time.sleep(2)

# 1-1) iframe 요소를 찾아 i_fream 변수에 할당
i_fream = driver.find_element(
    By.CSS_SELECTOR,
    "#_IframeBannerRight_tgtLREC"
)

# iframe 안으로 전환
driver.switch_to.frame(i_fream)

# 1-2) iframe 내부의 새로운 HTML 문서에서
#      class명이 content_info인 요소를 content 변수에 할당
content = driver.find_element(
    By.CLASS_NAME,
    "content_info"
)

# 1-3) content 안에 있는 자식 태그 3개의 내용 출력
children = content.find_elements(
    By.XPATH,
    "./*"
)

for child in children:
    print(child.text)

# 작업이 끝났으면 원래 HTML 문서로 돌아오기
driver.switch_to.default_content()
```
